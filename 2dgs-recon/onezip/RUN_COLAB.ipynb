```python
import glob, os, zipfile
CODE = "/content/onezip"; os.makedirs("/content/photos", exist_ok=True)
for p in glob.glob("/content/*.zip"):
    with zipfile.ZipFile(p) as zf:
        zf.extractall(CODE if "run_recon.py" in zf.namelist() else f"/content/photos/{os.path.basename(p)[:-4]}")
assert os.path.exists(f"{CODE}/run_recon.py"), "microtom_recon_onezip_v0.2.2.zip 을 /content 에 올리세요"
print("code:", sorted(os.listdir(CODE)))
```
